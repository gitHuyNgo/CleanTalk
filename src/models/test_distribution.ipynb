```python
import pandas as pd
```

```python
RESULT_PATH = './data/output/submit.csv'
#RESULT_PATH = './data/raw/train.csv'
labels = ['toxic', 'severe_toxic', 'obscene', 'threat', 'insult', 'identity_hate']
```

```python
df=pd.read_csv(RESULT_PATH)
df.head()
```

```python
def combs_to_label(combs):
    labels_name = [labels[i] for i, bit in enumerate(combs) if bit == "1"]
    return ",".join(labels_name) if labels_name else "non-toxic"
df["combs"] = df[labels].astype(str).agg("".join, axis=1)
result = df["combs"].value_counts().reset_index()
result.columns = ["combs", "count"]
result["labels"] = result["combs"].apply(combs_to_label)
print(result)
```
